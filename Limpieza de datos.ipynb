# Limpieza padron_poblacion


```python
import re

carpeta = "TablasOriginales"

padron_poblacion =  pd.read_excel(carpeta + "/padron_poblacion.xlsX")

padron_poblacion.columns = ['codigo', 'edad', 'casos', 'porcentaje', 'porcentaje_acumulado']
padron_poblacion.columns


edad_index = []
total_index = []
areas = []
codigos = []
counter = 0
counter_total = 0
for i in range(len(padron_poblacion['edad'].tolist())): # saco los indices de fila donde aparece Edad 
    if padron_poblacion['edad'].tolist()[i] == 'Edad':
        counter += 1
        edad_index.append(i+1)
        areas.append(padron_poblacion['casos'].tolist()[i-2])
        if(padron_poblacion['edad'].tolist()[i-2] != 'RESUMEN'):
            cod = re.search('AREA # (.*)',padron_poblacion['edad'].tolist()[i-2])
            cod = cod.group(1)
            codigos.append(cod)


for i in range(len(padron_poblacion['edad'].tolist())): # saco los indices de fila donde aparece Total
    if padron_poblacion['edad'].tolist()[i] == 'Total':
        total_index.append(i)
        counter_total += 1

if  counter != counter_total:
    print("Error: Tamaños de indexacion diferentes")
    sys.exit()


df_list = []
for index in range(counter-1):
    edad = padron_poblacion['edad'][edad_index[index]: total_index[index]].tolist()
    casos = padron_poblacion['casos'][edad_index[index]: total_index[index]].tolist()
    porcentaje = padron_poblacion['porcentaje'][edad_index[index]: total_index[index]].tolist()
    porcentaje_acumulado = padron_poblacion['porcentaje_acumulado'][edad_index[index]: total_index[index]].tolist()
    
    data = {
        'codigo': codigos[index],
        'edad': edad,
        'casos': casos,
        'porcentaje': porcentaje,
        'porcentaje_acumulado': porcentaje_acumulado
    }

    df_ = pd.DataFrame(data)

    ruta_subcarpeta = os.path.join('TablasModificadas', 'Padrones')

    os.makedirs(ruta_subcarpeta, exist_ok=True)

    ruta_archivo = os.path.join(ruta_subcarpeta, f'{codigos[index]}padron_poblacion_{areas[index]}.csv')
    df_.to_csv(ruta_archivo, index=False)
    
    df_list.append(df_)

```

Creacion de Departamentos 

Notar que los departamentos Chascomus, Rio Grande y Ushuaia, tienen codigos distintos en "Padrones" y "Datos por departamento actividad  y sexo", otra inconsistencia mas entre los datos

```python
import pandas as pd
import os
import re
from unidecode import unidecode

df = pd.DataFrame(columns=['ID','Nombre', 'Provincia', 'Grupo_Jardin', 'Grupo_Primaria', 'Grupo_Secundaria', 'Grupo_Adulto', 'Grupo_Mayores'])

columnas_deseadas = ["in_departamentos","departamento","provincia"]
df_departamentos = pd.read_csv("TablasOriginales/Datos_por_departamento_y_actividad.csv", usecols=columnas_deseadas)
df_departamentos = df_departamentos.drop_duplicates()

direccion = "TablasModificadas\Padrones"
i = 0
for file in os.listdir(direccion):
    df_test = pd.read_csv(direccion +"/"+ file, index_col=False)
    nombre = re.search('padron_poblacion_(.*).csv', file)
    nombre = nombre.group(1)
    if(file[0] == '0'):
        codigo = file[1:5]
    else:
        codigo = file[:5]
    provincia = df_departamentos.loc[df_departamentos['in_departamentos'] == int(codigo), 'provincia']
    g_jardin = df_test.loc[df_test['edad'] < 6, 'casos'].sum()
    g_prim = df_test.loc[(df_test['edad'] >= 6) & (df_test['edad'] < 13), 'casos'].sum()
    g_sec = df_test.loc[(df_test['edad'] >= 13) & (df_test['edad'] < 19), 'casos'].sum()
    g_adult = df_test.loc[(df_test['edad'] >= 19) & (df_test['edad'] < 60), 'casos'].sum()
    g_mayores = df_test.loc[df_test['edad'] > 60, 'casos'].sum()
    provincia = provincia.values
    if len(provincia) != 0:
        provincia = provincia[0]
    else:
        provincia = 'Null'
    data = {
        'ID' : codigo,
        'Nombre' : nombre,
        'Provincia' : provincia,
        'Grupo_Jardin' : g_jardin,
        'Grupo_Primaria': g_prim,
        'Grupo_Secundaria' : g_sec,
        'Grupo_Adulto' : g_adult,
        'Grupo_Mayores' : g_mayores
    }

    df.loc[len(df)] = data

tdf = 'Tierra Del Fuego'
bs = 'Buenos Aires'
df.loc[(df['Nombre'] == 'Ushuaia') & (df['Provincia'] == 'Null'), 'Provincia'] = tdf
df.loc[(df['Nombre'] == 'Río Grande') & (df['Provincia'] == 'Null'), 'Provincia'] = tdf
df.loc[(df['Nombre'] == 'Chascomús') & (df['Provincia'] == 'Null'), 'Provincia'] = bs

df['Nombre'] = df['Nombre'].apply(unidecode)

ruta = os.path.join('TablasModificadas', 'DER_Implement')
os.makedirs(ruta, exist_ok=True)
ruta_final = os.path.join(ruta, 'Departamento.csv')
df.to_csv(ruta_final, index=False)
```

```python
import pandas as pd
import duckdb as dd

folder = 'TablasModificadas/DER_Implement/'
tipos = pd.read_csv(folder + 'Es_de_tipo.csv')
establecimientos = pd.read_csv(folder + 'Establecimiento_Educativo.csv')
dep = pd.read_csv(folder + 'Departamento.csv')

consulta = """
    SELECT id_departamento, tipo_establecimiento, COUNT(*) AS cant_tipos
    FROM establecimientos LEFT JOIN tipos ON establecimientos.Cueanexo = tipos.Cueanexo
    WHERE tipo_establecimiento != 'Formación_Profesional' AND tipo_establecimiento != 'Formación_Profesional_INET'
    GROUP BY id_departamento, tipo_establecimiento
    ORDER BY id_departamento
"""
res = dd.sql(consulta).df()

consulta2 = """
    SELECT 
        id_departamento,
        SUM(CASE WHEN tipo_establecimiento = 'Jardín_maternal' THEN cant_tipos END) AS Jardin_Maternal,
        SUM(CASE WHEN tipo_establecimiento = 'Jardín_de_infantes' THEN cant_tipos END) AS Jardin_Infantes,
        SUM(CASE WHEN tipo_establecimiento = 'Primario' THEN cant_tipos END) AS Primarias,
        SUM(CASE WHEN tipo_establecimiento = 'Secundario' THEN cant_tipos END) AS Secundarias,
        SUM(CASE WHEN tipo_establecimiento = 'Secundario_INET' THEN cant_tipos END) AS Secundarias_INET,
        SUM(CASE WHEN tipo_establecimiento = 'SNU' THEN cant_tipos END) AS SNU,
        SUM(CASE WHEN tipo_establecimiento = 'SNU_INET' THEN cant_tipos END) AS SNU_INET 
    FROM res 
    GROUP BY id_departamento
"""
res = dd.sql(consulta2).df()

consulta3 = """
    SELECT 
        Provincia, 
        Nombre AS Departamento, 
        (Jardin_Maternal + Jardin_Infantes) AS Jardines,
        Grupo_Jardin AS Poblacion_Jardin,
        Primarias,
        Grupo_Primaria AS Poblacion_Primaria,
        (Secundarias + Secundarias_INET) AS Secundarias,
        Grupo_Secundaria AS Poblacion_Secundaria
    FROM res LEFT JOIN dep ON id_departamento = ID
    ORDER BY Provincia, Primarias
"""
res = dd.sql(consulta3).df()
res
```
